```python
import subprocess
subprocess.run(["date"])
subprocess.run(["sleep", "2"])
result = subprocess.run(["ls", "this file does not exist"])
print(result.returncode)
```

```python
print("Subprocess result without capture output")
result = subprocess.run(["host", "8.8.8.8"])
print(result)

print("\nSubprocess result with capture output")
result = subprocess.run(["host", "8.8.8.8"], capture_output=True)
print(f"result: {result}")
print(f"return code: {result.returncode}")
print(f"stdout: {result.stdout}")
print(f"stdout (decoded): {result.stdout.decode()}")
print(f"stdout (decoded and split): {result.stdout.decode().split()}")
```

```python
result = subprocess.run(["rm", "does_not_exist"])
print(f"result without capture output: {result}")

result = subprocess.run(["rm", "does_not_exist"], capture_output=True)
print(f"result with capture output: {result}")

result = subprocess.run(["rm", "does_not_exist"], capture_output=True)
print(f"return code: {result.returncode}")
print(f"stdout: {result.stdout}")
print(f"stderr: {result.stderr}")
```
