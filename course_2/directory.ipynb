```python
import os
from pathlib import Path
import datetime
```

```python
# listing all files and directories in the current directory
dir = "myenv"

for name in os.listdir(dir):
    full_path = os.path.join(dir, name)
    if os.path.isdir(full_path):
        print(f"{name} is a directory")
    else:
        print(f"{name} is a file")
```

```python
# listing all files and directories in the parent directory
parent_dir = os.path.dirname(os.getcwd())
print(f"Parent directory: {parent_dir}")

for name in os.listdir(parent_dir):
    full_path = os.path.join(parent_dir, name)
    if os.path.isdir(full_path):
        print(f"{name} is a directory")
    else:
        print(f"{name} is a file")
```

```python
# moving the file 
destination = os.path.join(os.path.join(os.getcwd()), "test")
if not os.path.exists(destination):
    os.mkdir(destination)

src_file = os.path.join(os.getcwd(), "test.py")
dst_file = os.path.join(destination, "test.py")

os.rename(src_file, dst_file)
```

```python
# moving the file back to the original location using pathlyb
from pathlib import Path
destination = Path(os.getcwd()) / "test"
src_file = Path(destination) / "test.py"
dst_file = Path(os.getcwd()) / "test.py"
src_file.rename(dst_file)
os.rmdir(destination)
```

```python
#get the modified time of the file
file_path = os.path.join(os.getcwd(), "test.py")
modified_time = os.path.getmtime(file_path)
print("{:.10s}".format(str(datetime.datetime.fromtimestamp(modified_time))))
```
